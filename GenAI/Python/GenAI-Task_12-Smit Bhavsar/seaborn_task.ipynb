Task 1: Relational Plot

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

dataframe = pd.read_csv('student_data.csv')

sns.relplot(data=dataframe, x='age', y='studytime', kind='line', hue='sex')
```

Task 2: Line Plot as Sctter & Facet

```python
sns.lineplot(data=dataframe, x='age', y='studytime', hue='sex')
```

```python
sns.scatterplot(data=dataframe, x='age', y='studytime', hue='sex')
```

```python
sns.relplot(data=dataframe, x='age', y='studytime', kind='line', hue='sex', col='sex')
```

Task 3: Distribution Plot

```python
sns.histplot(data=dataframe, x='G3', discrete=True)
plt.show()

sns.kdeplot(data=dataframe, x='G3')
plt.show()

sns.rugplot(data=dataframe, x='G3', height=0.5)
plt.show()

sns.histplot(data=dataframe, x='G3', discrete=True, kde=True)
plt.show()
```

Task 4: Bivariate Distribution Plots

```python

sns.histplot(data=dataframe, x='age', y='G3', bins=10, cbar=True, cmap='Blues')
plt.show()

plt.figure(figsize=(8, 6))
sns.kdeplot(
    data=dataframe,
    x='age',
    y='G3',
    fill=True,
    cmap='mako',
)
plt.show()
```

Task 5: Matrix Plot

```python
grade_data = dataframe[['G1', 'G2', 'G3']]

sns.pairplot(grade_data, diag_kind='hist')
plt.show()

correlation_matrix = grade_data.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Matrix of Student Grades')
plt.show()
```

Task 6: Categorical Plots

```python
sns.barplot(data=dataframe, x='sex', y='G3')
plt.title('Average Final Grade by Sex')
plt.xlabel('Sex')
plt.ylabel('Average Final Grade (G3)')
plt.show()

sns.boxplot(data=dataframe, x='sex', y='G3')
plt.title('Final Grade Distribution by Sex')
plt.xlabel('Sex')
plt.ylabel('Final Grade (G3)')
plt.show()

sns.violinplot(data=dataframe, x='sex', y='G3')
plt.title('Final Grade Distribution by Sex')
plt.xlabel('Sex')
plt.ylabel('Final Grade (G3)')
plt.show()

sns.countplot(data=dataframe, x='sex')
plt.title('Student Count by Sex')
plt.xlabel('Sex')
plt.ylabel('Number of Students')
plt.show()
```

Task 7: Regression Plots

```python
sns.regplot(data=dataframe, x='age', y='G3', scatter_kws={'alpha': 0.5})
plt.show()

sns.lmplot(data=dataframe, x='age', y='G3', hue='sex', height=6, aspect=1.2)
plt.show()
```

Task 8: Multi-Plots and Figure-Level Plots

```python
facet_grid = sns.FacetGrid(dataframe, col='sex').map(plt.scatter, 'age', 'G3')
facet_grid.set_axis_labels('Age', 'Final Grade (G3)')
plt.show()

rel_plot = sns.relplot(
    data=dataframe,
    x='age',
    y='G3',
    hue='sex',
    kind='scatter',
    height=5,
)
plt.show()

cat_plot = sns.catplot(
    data=dataframe,
    x='sex',
    y='G3',
    kind='box',
    height=5,
)
plt.show()

dist_plot = sns.displot(
    data=dataframe,
    x='G3',
    hue='sex',
    kind='hist',
    multiple='layer',
    element='step',
    stat='density',
    common_norm=False,
    bins=11,
    binrange=(0, 22),
    height=5,
)
plt.show()
```
