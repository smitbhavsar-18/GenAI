Task 1: Line Plot

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

dataframe = pd.read_csv('Coffe_sales.csv')
print()
month_order = [
    'Jan',
    'Feb',
    'Mar',
    'Apr',
    'May',
    'Jun',
    'Jul',
    'Aug',
    'Sep',
    'Oct',
    'Nov',
    'Dec',
]

dataframe['Month_name'] = pd.Categorical(
    dataframe['Month_name'], categories=month_order, ordered=True
)

grouped_data = dataframe['Month_name'].value_counts().sort_index()
x = grouped_data.index
y = grouped_data.values


plt.figure(figsize=(10, 6))
plt.plot(x, y, color='red', marker='*', linewidth=2, markersize=10)
plt.grid()
plt.title('Total Sales by Month')
plt.xlabel('Month')
plt.ylabel('Total Money')
plt.show()
```

Task 2: Scatter Plot

```python
plt.figure(figsize=(10, 6))
plt.scatter(x, y,color='red', marker='*')
plt.grid()
plt.title('Total Sales by Month')
plt.xlabel('Month')
plt.ylabel('Total Money')
plt.show()
```

Task 3 : Bar Plot

```python
plt.figure(figsize=(10, 6))
plt.bar(x, y,color='red')
plt.grid()
plt.title('Total Sales by Month')
plt.xlabel('Month')
plt.ylabel('Total Money')
plt.show()

plt.figure(figsize=(10, 6))
plt.bar(y, x,color='red')
plt.grid()
plt.title('Total Sales by Month')
plt.xlabel('Month')
plt.ylabel('Total Money')
plt.show()
```

Task 4: Multiple Bar Chart

```python
coffee_sales = pd.crosstab(dataframe['Month_name'], dataframe['coffee_name'])

bar_width = 1 / len(coffee_sales.columns)
month_positions = np.arange(len(month_order))

plt.figure(figsize=(15, 7))
for number, coffee in enumerate(coffee_sales.columns):
    plt.bar(
        month_positions + number * bar_width,
        coffee_sales[coffee],
        width=bar_width,
        label=coffee,
    )

label_positions = month_positions + bar_width * (len(coffee_sales.columns) - 1) / 2
plt.xticks(label_positions, month_order)
plt.title('Number of Coffee Sales by Month')
plt.xlabel('Month')
plt.ylabel('Number of Sales')
plt.grid()
plt.legend(title='Coffee Type')
plt.show()
```

Task 5: Stacked Bar Chart

```python
coffee_sales = pd.crosstab(dataframe['Month_name'], dataframe['coffee_name'])
month_positions = np.arange(len(month_order))

print(month_positions)

plt.figure(figsize=(15, 7))
for coffee in coffee_sales.columns:
    plt.bar(
        month_positions,
        coffee_sales[coffee],
        label=coffee,
    )

plt.xticks(month_positions, month_order)
plt.title('Stacked Coffee Sales by Month')
plt.xlabel('Month')
plt.ylabel('Number of Sales')
plt.grid()
plt.legend(title='Coffee Type')
plt.show()
```

Task 6: Histogram (Marks Distribution)

```python
marks_dataframe = pd.DataFrame(
    {'Marks': [45, 52, 58, 61, 63, 65, 67, 68, 70, 72, 74, 75, 76, 78, 80, 81, 82, 83, 84, 85, 87, 88, 89, 90, 92, 94, 96, 49, 55, 77]}
)

marks = marks_dataframe['Marks']
bin_edges = np.histogram_bin_edges(marks, bins='auto')

plt.figure(figsize=(10, 6))
plt.hist(marks, bins=bin_edges, color='teal', edgecolor='black')
plt.title(f'Marks Distribution ({len(bin_edges) - 1} bins)')
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

Task 7: Pie Chart (Coffee Market Share)

```python
coffee_market_share = dataframe['coffee_name'].value_counts()

plt.figure(figsize=(9, 9))
plt.pie(
    coffee_market_share,
    labels=coffee_market_share.index,
    autopct='%1.1f%%',
)
plt.title('Market Share by Coffee Type')
plt.axis('equal')
plt.show()
```
